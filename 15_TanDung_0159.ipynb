```python
class NhanVien:
    def __init__(self, ma_nv, phong_ban, luong_cb):
        self.ma_nv = ma_nv
        self.phong_ban = phong_ban
        self.luong_cb = float(luong_cb)
        
    def tinh_luong(self):
        return self.luong_cb
        
    def lay_ma_nv(self):
        return self.ma_nv 

# Lớp con tăng ca
class nhanvientangca(NhanVien):
    def tinh_luong(self): 
        return super().tinh_luong() * 1.2

# Lớp con chuyên cần
class nhanvienchuyenchuyen(NhanVien):
    def tinh_luong(self):
        return super().tinh_luong() + 500000

# ==== main ====
if __name__ == "__main__":
    ds_nhan_vien = [
        # Truyền đủ 3 tham số: mã, phòng, lương
        NhanVien("nv1", "Hanh chinh", 10000000), 
        nhanvientangca("nv2", "Ky thuat", 10000000),
        nhanvienchuyenchuyen("nv3", "San xuat", 10000000)
    ]
    
    for nv in ds_nhan_vien:
        print(f"ma nvien: {nv.lay_ma_nv()}, luong: {nv.tinh_luong():,.0f}")
```

```python
san_luong = [12, None, 8, 15, None, 10]
arr = np.array(san_luong, dtype=float).reshape(2, 3)
mean_val = np.nanmean(arr)
arr[np.isnan(arr)] = mean_val
print("Mảng sau khi reshape và xử lý NaN:")
print(arr)
phan_tu = arr[arr > 9]
min_val = np.min(phan_tu )
print("\nCác phần tử có giá trị > 9:", phan_tu )
print("Giá trị nhỏ nhất trong tập hợp trên:", min_val)
```

```python
df=pd.read_csv('nhan_su.csv')
print(df)
```

```python
ket_qua = df[
    (df['Ma_NV'].str.contains('NS', case=False, na=False)) & 
    (df['Nam_Cong_Tac'] >= 2021)
]
print(ket_qua)
```

```python
mean_thanh_tich = ket_qua['Thanh_Tich'].mean()
ket_qua['Thanh_Tich'] = ket_qua['Thanh_Tich'].fillna(mean_thanh_tich)
print(mean_thanh_tich)
```

```python
pivot_table_res = ket_qua.pivot_table(
    index='Phong_Ban',     
    columns='Nam_Cong_Tac', 
    values='Thanh_Tich',    
    aggfunc='sum',         
    fill_value=0
)
print(pivot_table_res)
```

```python
import sqlite3
conn = sqlite3.connect('HeThong.db')
pivot_table_res.to_sql('bang_luong_2024', conn, if_exists='replace')
query = "SELECT * FROM bang_luong_2024 LIMIT 5"
df_check = pd.read_sql_query(query, conn)
conn.close()
print("5 dòng đầu tiên của bảng bang_luong_2024:")
print(df_check)
```
